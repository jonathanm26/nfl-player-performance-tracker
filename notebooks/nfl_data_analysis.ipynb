# NFL Player Data Analysis

This notebook explores NFL player statistics using Pandas and visualizations.

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/nfl_player_stats.csv')
df.head()
```

```
# Basic dataset information
df.info()
```

```
# Top players by estimated fantasy points
top_players = df.sort_values('fantasy_points_est', ascending=False)
top_players[['player', 'team', 'position', 'fantasy_points_est']].head(10)
```

```
# Plot fantasy points
top_players.head(10).plot(
    kind='bar',
    x='player',
    y='fantasy_points_est',
    figsize=(10,5)
)
plt.title('Top NFL Players by Estimated Fantasy Points')
plt.ylabel('Fantasy Points')
plt.show()
```
